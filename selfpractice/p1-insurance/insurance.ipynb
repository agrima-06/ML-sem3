```python
import numpy as np 
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt 
import warnings 
#print(sns.__version__)
warnings.filterwarnings('ignore')  #Ignore warnings for cleaner output
```

```python
numeric_col = ['age', 'bmi', 'children', 'charges'] #to check the numeric columns in the dataset (khud se select kra gya hai to check teh cols)
for col in numeric_col:
    plt.figure(figsize=(6,4))#Create a 6×4 graph area #its helps decide the height and width of the graph area
    sns.histplot(data[col],kde = True,bins = 20)#reate a histogram showing the distribution of data in col 
    #bins = 20 means the data is divided into 20 distributions
    #kde = True means the kernel density estimate is shown in the histogram 
    #It adds a smooth curve over your histogram showing the approximate distribution/shape of the data.
    
```

### Q1.

### Q13.

A and B are both 2x3 matrices. Compute `A @ B.T` and `A.T @ B`. Print the shape of each result and explain in a comment why both multiplications are legal while `A @ B` is not.

```python
A = np.array([[1, 2, 3],
              [4, 5, 6]])
B = np.array([[7, 8, 9],
              [1, 0, 2]])
```

````python
# Q13
A = np.array([[1, 2, 3],
              [4, 5, 6]])
B = np.array([[7, 8, 9],
              [1, 0, 2]])

print(f"A shape: {A.shape}")
print(f"B shape: {B.shape}")

# A @ B.T
result1 = A @ B.T
print(f"\nA @ B.T:")
print(result1)
print(f"Shape: {result1.shape}")

# A.T @ B
result2 = A.T @ B
print(f"\nA.T @ B:")
print(result2)
print(f"Shape: {result2.shape}")

# Why A @ B is not legal:
# A is (2, 3) and B is (2, 3)
# For matrix multiplication A @ B, the number of columns in A (3) must equal the number of rows in B (2)
# But 3 ≠ 2, so A @ B is not allowed
# However, A @ B.T: A is (2, 3) and B.T is (3, 2), so 3 == 3 ✓
# And A.T @ B: A.T is (3, 2) and B is (2, 3), so 2 == 2 ✓
````

### Q14.

Build a complete multi-step pipeline. Starting from the two arrays below: convert them into column vectors, join them into a single two-column table, attach a third column holding each row's total, then multiply the three-column table by `w = [0.5, 0.2, 0.3]` to obtain one score per row. Finally use `np.where` to mark each score as 1 if it is above the average score and 0 otherwise. Report the shape of every intermediate array.

```python
hours = np.array([2, 8, 5, 10, 3, 7])
attendance = np.array([55, 92, 70, 98, 60, 85])
```

````python
# Q14
hours = np.array([2, 8, 5, 10, 3, 7])
attendance = np.array([55, 92, 70, 98, 60, 85])

# Convert to column vectors
hours_col = hours.reshape(-1, 1)
attendance_col = attendance.reshape(-1, 1)
print(f"Hours column shape: {hours_col.shape}")
print(f"Attendance column shape: {attendance_col.shape}")

# Join into two-column table
two_col_table = np.hstack([hours_col, attendance_col])
print(f"\nTwo-column table shape: {two_col_table.shape}")
print(two_col_table)

# Attach third column: row totals
row_totals = (hours_col + attendance_col)
three_col_table = np.hstack([two_col_table, row_totals])
print(f"\nThree-column table shape: {three_col_table.shape}")
print(three_col_table)

# Multiply by weights
w = np.array([0.5, 0.2, 0.3])
scores = three_col_table @ w
print(f"\nScores (after multiplying by weights) shape: {scores.shape}")
print(f"Scores: {scores}")

# Mark each score as 1 if above average, 0 otherwise
avg_score = np.mean(scores)
print(f"\nAverage score: {avg_score}")

marks = np.where(scores > avg_score, 1, 0)
print(f"Marks (1 if above average, 0 otherwise) shape: {marks.shape}")
print(f"Marks: {marks}")
````

### Q16.

Take the array below and create two 1D versions of it — one with `flatten()` and one with `ravel()`. Change the first element of each, then print the original matrix. Explain the difference you observe in a comment.

```python
m = np.array([[4, 8, 15],
              [16, 23, 42]])
```

````python
# Q16
m = np.array([[4, 8, 15],
              [16, 23, 42]])

print("Original matrix:")
print(m)

# Using flatten()
flat_copy = m.flatten()
print(f"\nflat_copy using flatten(): {flat_copy}")
flat_copy[0] = 999
print(f"After changing first element to 999: {flat_copy}")
print(f"Original matrix after flatten() modification:")
print(m)

# Using ravel()
flat_view = m.ravel()
print(f"\nflat_view using ravel(): {flat_view}")
flat_view[0] = 999
print(f"After changing first element to 999: {flat_view}")
print(f"Original matrix after ravel() modification:")
print(m)

# Difference explanation:
# flatten() returns a COPY of the array (independent)
# ravel() returns a VIEW of the array when possible (not independent)
# The second modification survived because ravel() creates a view that still references the original array
````

### Q18.

Create two random 3x3 matrices A and B, compute `A @ B` and `B @ A`, and check whether the two results are equal. State your conclusion about matrix multiplication in a comment.

````python
# Q18
np.random.seed(42)
A = np.random.rand(3, 3)
B = np.random.rand(3, 3)

print("Matrix A:")
print(A)
print("\nMatrix B:")
print(B)

# Compute A @ B
AB = A @ B
print("\nA @ B:")
print(AB)

# Compute B @ A
BA = B @ A
print("\nB @ A:")
print(BA)

# Check equality
are_equal = np.allclose(AB, BA)
print(f"\nAre A @ B and B @ A equal? {are_equal}")

# Conclusion about matrix multiplication
print("\nConclusion: Matrix multiplication is NOT commutative.")
print("In general, A @ B ≠ B @ A for arbitrary matrices A and B.")
````

### Q20.

Start from `arr = np.arange(1, 11)`. Create one variable using a plain slice `arr[2:6]` and another using `arr[2:6].copy()`. Change the first element of each to 999, then print the original array and explain in a comment which change survived and why.

````python
# Q20
arr = np.arange(1, 11)
print("Original array:")
print(arr)

# Create a plain slice
plain_slice = arr[2:6]
print(f"\nPlain slice arr[2:6]: {plain_slice}")

# Create a copy
copied_slice = arr[2:6].copy()
print(f"Copied slice arr[2:6].copy(): {copied_slice}")

# Change first element of each to 999
plain_slice[0] = 999
copied_slice[0] = 999

print(f"\nAfter changing first element of each to 999:")
print(f"Plain slice: {plain_slice}")
print(f"Copied slice: {copied_slice}")

print(f"\nOriginal array:")
print(arr)

# Explanation:
# The plain slice creates a VIEW into the original array, so changes affect the original.
# The copied slice creates an INDEPENDENT copy, so changes don't affect the original.
# Only the plain slice change survived in the original array.
````
